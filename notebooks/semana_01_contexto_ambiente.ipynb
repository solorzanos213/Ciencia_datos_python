## Verificación de carpetas

Busca la raíz del proyecto a partir de la ubicación actual para que funcione al ejecutar VS Code desde la carpeta raíz o desde `notebooks/`.


```python
actual = Path.cwd().resolve()
BASE_DIR = next((p for p in (actual, *actual.parents) if (p / "README.md").exists() and (p / "notebooks").exists()), None)
if BASE_DIR is None:
    raise FileNotFoundError("Abre el proyecto ciencia-datos-python desde VS Code antes de ejecutar.")
carpetas = ["data/raw", "data/processed", "notebooks", "outputs", "docs", "scripts"]
for carpeta in carpetas:
    ruta = BASE_DIR / carpeta
    ruta.mkdir(parents=True, exist_ok=True)
    print("Verificada:", ruta)
```
